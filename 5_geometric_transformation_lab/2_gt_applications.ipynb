To further illustrate the advantage of cubic interpolation, let us now plot the signal-to-noise ratio of the three interpolators when rotating an image $n$ times by $35^{\circ}$ and then $n$ times by $-35^{\circ}$.

```python
# Function that computes the signal-to-noise ratio between two images
def SNR(img, img_ref):
    return 10 * np.log10(
        np.sum(img.astype(np.float64) ** 2)
        / np.sum((img_ref.astype(np.float64) - img.astype(np.float64)) ** 2)
    )


# Rotate images up to 10 times by 35 degrees and back
alpha = 35
ns = np.arange(1, 11, dtype=int)
snr_nn = [
    SNR(rotate_n(rotate_n(alsace, alpha, n, 0), -alpha, n, 0), alsace) for n in ns
]
snr_lin = [
    SNR(rotate_n(rotate_n(alsace, alpha, n, 1), -alpha, n, 1), alsace) for n in ns
]
snr_cub = [
    SNR(rotate_n(rotate_n(alsace, alpha, n, 3), -alpha, n, 3), alsace) for n in ns
]
# Display graphs
plt.close("all")
plt.figure()
plt.plot(ns, snr_nn, label="nn")
plt.plot(ns, snr_lin, label="linear")
plt.plot(ns, snr_cub, label="cubic")
plt.legend()
plt.title("SNR of images rotated n times by $35^{\\circ}$ and back")
plt.grid()
plt.ylabel("SNR [dB]")
plt.xlabel("n")
plt.show()
```

In the next cell, we will implement the function `transform_polar` to create the image `polar`. In it, each pixel is a discretized element of the polar space $(\rho, \theta)$, with **$\rho$ being represented in the $y$-axis and $\theta$ in the $x$-axis**. Since our clock spans the entire image, our radius can be defined, in pixels, as half the image size. Therefore, we will take `nr` as half of the average between `nx` and `ny`. Furthermore, we will set the angular resolution to $1^\circ$ and set `na` to $720$. 

We then call your `transform_polar_coordinates` function to get the `x` and `y` corresponding to every `rho` and `theta` using [`np.fromfunction`](https://numpy.org/doc/stable/reference/generated/numpy.fromfunction.html) and then interpolate the pixel values at `x` and `y` in the original to create the output `polar` output image. We then flip it for convenient visualization.

The function `transform_polar` takes as inputs:
 * `img`: The original image to transform,  
 * `kc`: a list  of $2$ elements that indicates the center offset in $x$ (`kc[0]`) and $y$ (`kc[1]`), 
 * `kr`: a list  of $2$ elements that indicates the radius offset (`kr[0]`) and scaling (`kr[1]`) in the form $\rho(d) = k_r[0] + d\cdot k_r[1]$, where $d$ is the distance from the center, and
 * `ka`: a list  of $2$ elements that, as `kr` indicates the angle offset in the form $\alpha(d) = k_a[0] + d\cdot k_a[1]$.
 
The coefficients `kc`, `kr`, and `ka` are initialized to neutral values by default; we will modify them in the following cell.

Make sure to understand all the code in the next cell!

```python
# Define neutral value coefficients (no offset: 0 and unit scale: 1)
kc = [0, 0]
kr = [0, 1]
ka = [0, 1]


# Define transformation combining your coordinates mapping and interpolation to reconstruct the image
def transform_polar(img, kc, kr, ka):
    nx, ny = img.shape
    # Define image with the radius in the y-axis set as the average radius of the input image and the angle on the x-axis
    nr = (nx + ny) // 4
    na = 720
    polar = np.zeros((nr, na))

    # offset transformation center
    cx = nx // 2 + kc[0]
    cy = ny // 2 + kc[1]

    # define utility first order development for offset and scaling of rho, theta
    def fod(k, x):
        return k[0] + k[1] * x

    # Extract coordinate positions in original image
    x, y = np.fromfunction(
        lambda rho, theta: transform_polar_coordinates(
            (cx, cy), fod(kr, rho), fod(ka, theta)
        ),
        shape=(nr, na),
    )
    # Extract corresponding values
    polar = ndimage.map_coordinates(img, (x, y), order=3, mode="constant").reshape(
        polar.shape
    )
    # Flip for easier watch reading
    polar = np.flip(polar)

    return polar
```

It is time to introduce our time reading utility to have a read through it. We compute the sum of intensity values along the radius and then infer the time from the index at which the intensity minima occur.

```python
# function that returns a string indicating the time it reads from the transformed image
def read_time(image, kc, kr, ka):
    # Get the polar image and its shape
    polar = transform_polar(image, kc, kr, ka)
    nr, na = polar.shape
    # Sum along the vertical axis
    intensity = np.sum(polar, axis=0)

    # Locate the index of the minimum, this is where the minute hand lies (since it is the most prominent black area)
    # Then, remove the black area within the margin around the minute hand by setting it to a very high value
    proj = intensity.copy()
    margin = 15
    minutes = np.argmin(proj)
    # nr accounts for the previous summation of pixels
    proj[np.max([0, minutes - margin]) : np.min([na, minutes + margin])] = 255 * nr
    # Locate the index of the new minimum, this is now where the hour hand lies
    hours = np.argmin(proj)

    # Thanks to your definitions of kc, kr, ka there should now exist a linear relationship between the angle and the time
    return (
        (intensity, polar, (nr, na, margin)),
        (
            hours,
            minutes,
            f"{int(np.floor(12 * hours / na))}h{int(np.round(60 * minutes / na))}",
        ),
    )
```
